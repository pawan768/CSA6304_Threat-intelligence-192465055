```python
events = [
    {"user": "Ravi", "status": "failed", "source": "10.0.0.1"},
    {"user": "Ravi", "status": "failed", "source": "10.0.0.2"},
    {"user": "Anu", "status": "success", "source": "10.0.0.1"},
    {"user": "Ravi", "status": "success", "source": "10.0.0.1"},
    {"user": "Anu", "status": "failed", "source": "10.0.0.3"}
]

# Search filter
print("Failed login events:")
filtered = [e for e in events if e["status"] == "failed"]

for event in filtered:
    print(event)

# Stats count by user
print("\nEvent count by user:")
counts = {}

for event in events:
    user = event["user"]
    counts[user] = counts.get(user, 0) + 1

for user, count in counts.items():
    print(user, ":", count)
```

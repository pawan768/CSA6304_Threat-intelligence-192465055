```python
def classify_control(description):
    text = description.lower()

    domains = {
        "Organizational": ["policy", "risk management", "supplier",
                           "incident management"],
        "People": ["awareness", "training", "employee", "personnel"],
        "Physical": ["door", "cctv", "physical access", "server room"],
        "Technological": ["firewall", "encryption", "antivirus",
                          "access control", "network"]
    }

    for domain, keywords in domains.items():
        if any(word in text for word in keywords):
            return

    return "Domain not identified"


description = input("Enter security control description: ")
print("ISO 27001 Domain:", classify_control(description))
```

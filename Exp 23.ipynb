```python
# Known attack signatures
signatures = ["SQL Injection", "Port Scan", "Malware"]

# Trusted source whitelist
whitelist = ["192.168.1.10"]

traffic = [
    ("192.168.1.10", "SQL Injection"),
    ("192.168.1.20", "Malware"),
    ("192.168.1.30", "Normal Traffic")
]

for ip, data in traffic:

    if ip in whitelist:
        print(ip, "- Allowed (Trusted Source)")

    elif any(signature in data for signature in signatures):
        print(ip, "- BLOCKED (Malicious Traffic)")

    else:
        print(ip, "- Allowed (Normal Traffic)")
```

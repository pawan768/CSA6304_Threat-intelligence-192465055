```python
scenario = input("Enter business scenario: ").lower()

if "bank" in scenario or "financial" in scenario or "payment" in scenario:
    print("Threat Intelligence Area: Banking Fraud Prevention")
elif "hospital" in scenario or "healthcare" in scenario or "patient" in scenario:
    print("Threat Intelligence Area: Healthcare Data Protection")
elif "online shopping" in scenario or "e-commerce" in scenario or "customer orders" in scenario:
    print("Threat Intelligence Area: E-commerce Fraud Prevention")
else:
    print("Threat Intelligence Area: General Cybersecurity")
```

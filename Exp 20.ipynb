```python
def classify_threat(description):
    text = description.lower()

    keywords = {
        "Phishing": ["fake email", "password", "fraudulent link", "phishing"],
        "DoS/DDoS": ["traffic flood", "server overload", "dos", "ddos"],
        "Malware": ["virus", "ransomware", "trojan", "malware"],
        "Man-in-the-Middle": ["intercept", "eavesdrop", "data interception", "mitm"]
    }

    for threat, words in keywords.items():
        for word in words:
            if word in text:
                return threat

    return "Unknown Threat"

description = input("Enter incident description: ")
result = classify_threat(description)

print("Detected Threat Category:", result)
```

```python
import os

# Sample logs to create if auth.log does not exist
sample_logs = """\
Jan 1 10:00:00 server sshd[123]: Failed password for basheer from 192.168.1.50 port 22 ssh2
Jan 1 10:01:00 server sshd[123]: Failed password for basheer from 192.168.1.50 port 22 ssh2
Jan 1 10:02:00 server sshd[123]: Failed password for basheer from 192.168.1.50 port 22 ssh2
Jan 1 10:03:00 server sshd[123]: Accepted password for basheer from 192.168.1.50 port 22 ssh2
"""

# Create a sample auth.log file if it doesn't exist
if not os.path.exists("auth.log"):
    with open("auth.log", "w") as f:
        f.write(sample_logs)
    print("Created a sample 'auth.log' file for demonstration purposes.\n")

user = input("Enter username: ")
failed = 0
success = False

try:
    with open("auth.log", "r") as file:
        for line in file:
            if user in line:
                if "Failed password" in line:
                    failed += 1
                elif "Accepted password" in line:
                    success = True

    if failed >= 3 and success:
        print("\nBrute Force Attack Detected!")
        print("Failed Attempts:", failed)
        print("Login Successful after failures.")
    else:
        print("\nNo Brute Force Attack Detected.")
except FileNotFoundError:
    print("\nError: 'auth.log' could not be found or created.")
```

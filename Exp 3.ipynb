```python
policy = input("Enter security policy text: ").lower()

elements = {
    "Scope": ["scope", "applies to", "coverage"],
    "Roles and Responsibilities": ["roles", "responsibilities"],
    "Enforcement": ["enforcement", "disciplinary action"],
    "Review Cycle": ["review cycle", "review annually", "review every year"],
    "Approval Authority": ["approved by", "approval authority"]
}

missing = []

for element, keywords in elements.items():
    if not any(word in policy for word in keywords):
        missing.append(element)

if missing:
    print("\nMissing Governance Elements:")
    for element in missing:
        print("-", element)
else:
    print("\nAll required governance elements are present.")
```

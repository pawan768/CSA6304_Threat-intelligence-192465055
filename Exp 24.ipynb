```python
# Current security configuration
current = {
    "Firewall": 1,
    "Antivirus": 1,
    "Password Length": 6,
    "Backup Frequency": 1,
    "MFA": 0
}

# Minimum required security baseline
baseline = {
    "Firewall": 1,
    "Antivirus": 1,
    "Password Length": 8,
    "Backup Frequency": 2,
    "MFA": 1
}

print("Security Baseline Report")
print("------------------------")

for control in baseline:

    if current[control] < baseline[control]:
        print(control, "- FAIL")
    else:
        print(control, "- PASS")
```
